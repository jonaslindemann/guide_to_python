````python
def newton_raphson(x0, f, f_p, eps=1e-6):

    diff = 1e300

    while diff > eps:
        x1 = x0 - f(x0)/f_p(x0)
        diff = abs(x1 - x0)
        x0 = x1

    return x0

if __name__ == "__main__":

    x = newton_raphson(1.0, f, f_p)

    print(x)
````

### Sökning i sorterade listor

````python
import bisect

sorterad_lista = [1, 3, 4, 7, 9, 12, 15, 18, 20]

# Hitta position där värdet skulle infogas
pos = bisect.bisect_left(sorterad_lista, 7)
print(f"Värdet 7 finns på position: {pos}")

# Kontrollera om ett värde finns i listan
def finns_i_sorterad_lista(lista, värde):
    pos = bisect.bisect_left(lista, värde)
    return pos < len(lista) and lista[pos] == värde

print(f"Finns 7 i listan? {finns_i_sorterad_lista(sorterad_lista, 7)}")
print(f"Finns 8 i listan? {finns_i_sorterad_lista(sorterad_lista, 8)}")
````

#### Counter - räkna förekomster

#### deque - dubbelriktad kö

````python
from collections import deque

# Skapa en kö
kö = deque()

# Lägg till element
kö.append("sist")        # Lägg till i slutet
kö.appendleft("först")   # Lägg till i början
print(f"Kö: {kö}")

# Ta bort element
första = kö.popleft()    # Ta från början
sista = kö.pop()         # Ta från slutet
print(f"Första: {första}, Sista: {sista}")

# Rotera elementen
kö.extend([1, 2, 3, 4, 5])
print(f"Före rotation: {kö}")
kö.rotate(2)  # Rotera 2 steg åt höger
print(f"Efter rotation: {kö}")
````

#### defaultdict - ordlista med standardvärden

````python
from collections import defaultdict

# Gruppera data efter kategori

studenter = [
    ("Alice", "Matematik"), ("Bob", "Fysik"), ("Charlie", "Matematik"),
    ("Diana", "Kemi"), ("Erik", "Fysik"), ("Fiona", "Matematik")
]

grupper = defaultdict(list)

for namn, ämne in studenter:
    grupper[ämne].append(namn)

for ämne, studenter in grupper.items():
    print(f"{ämne}: {studenter}")

# Räkna förekomster

räknare = defaultdict(int)
ord = ["äpple", "banan", "äpple", "citron", "banan", "äpple"]
for ord in ord:
    räknare[ord] += 1

print(f"Räknare: {dict(räknare)}")
````

### Heap - prioritetskö

````python
import heapq

# Skapa en heap
heap = [3, 1, 4, 1, 5, 9, 2, 6]
heapq.heapify(heap)  # Gör om listan till en heap
print(f"Heap: {heap}")

# Ta det minsta elementet
minsta = heapq.heappop(heap)
print(f"Minsta element: {minsta}")
print(f"Heap efter pop: {heap}")

# Lägg till nytt element
heapq.heappush(heap, 0)
print(f"Heap efter push(0): {heap}")

# Hitta de n minsta elementen
data = [23, 45, 12, 67, 34, 89, 11, 43]
tre_minsta = heapq.nsmallest(3, data)
tre_största = heapq.nlargest(3, data)
print(f"Tre minsta: {tre_minsta}")
print(f"Tre största: {tre_största}")
````

### Avancerade iterationsverktyg

````python
import itertools

# Kombinera flera sekvenser

lista1 = [1, 2, 3]
lista2 = ['a', 'b', 'c']
lista3 = [10, 20]

# Chain - slå ihop sekvenser

kombinerad = list(itertools.chain(lista1, lista2, lista3))
print(f"Chain: {kombinerad}")

# Product - kartesisk produkt

produkt = list(itertools.product([1, 2], ['a', 'b']))
print(f"Product: {produkt}")

# Zip_longest - som zip men fyller ut kortare sekvenser

parat = list(itertools.zip_longest(lista1, lista2, lista3, fillvalue=0))
print(f"Zip_longest: {parat}")

# Groupby - gruppera element

data = [('A', 1), ('A', 2), ('B', 3), ('B', 4), ('C', 5)]
for nyckel, grupp in itertools.groupby(data, key=lambda x: x[0]):
    print(f"Grupp {nyckel}: {list(grupp)}")
````

### Prestanda och användningstips

````python
import time
from collections import deque

# Jämför prestanda för list vs deque vid infogning i början

def benchmark_prepend(container_type, n=10000):
    container = container_type()
    start = time.time()

    for i in range(n):
        if hasattr(container, 'appendleft'):
            container.appendleft(i)
        else:
            container.insert(0, i)

    return time.time() - start

# Testa för mindre mängd data för demonstration

n = 10000
list_time = benchmark_prepend(list, n)
deque_time = benchmark_prepend(deque, n)

print(f"List prepend tid: {list_time:.4f}s")
print(f"Deque prepend tid: {deque_time:.4f}s")
print(f"Deque är {list_time/deque_time:.1f}x snabbare")
````

### Garanterad exekvering av kod efter undantag

````python
try:
    input_file = open("numbers.txt", "r")
    output_file = open("sums.txt", "w")

    lines = input_file.readlines()
    for line in lines:
        items = line.strip().split()
        numbers = []
        for item in items:
            numbers.append(int(item))

        output_file.write("%d\n" % sum(numbers))
finally:
    print("Stäng öppna filer.")
    input_file.close()
    output_file.close()
````

````python
def add_numbers(a: int, b: int) -> int:
    """Lägger ihop två heltal och returnerar summan."""
    try:
        return a + b
    except TypeError as e:
        print(f"Fel vid addition: {e}")
        return 0

result = add_numbers("3", 5)  # Detta kommer att ge en varning
print(result)
````

````python
try:
    result = add_numbers("3", 5)  # Detta kommer att ge en varning
except TypeError as e:
    print(f"Fel vid anrop av funktionen: {e}")
    result = 0

print(result)   
````

### Typehänvisningar för sammansatta typer

````python
from typing import List, Dict

# Lista av heltal
numbers: List[int] = [1, 2, 3, 4, 5]

# Uppslagslista med strängar som nycklar och heltal som värden
phone_book: Dict[str, int] = {
    "Arne": 47329823,
    "Bengt": 91238129,
    "Stina": 1928319,
    "Lena": 98129312
}
````

````python
def multiply_list(numbers: List[int], factor: int) -> List[int]:
    """Multiplicerar varje tal i en lista med en faktor."""
    return [n * factor for n in numbers]

numbers = [1, 2, 3, 4]
multiplied = multiply_list(numbers, 2)
print(f"Multiplicerad lista: {multiplied}")
````

````python
multiplied = multiply_list([1, 2, 3], "2")
````

### Typhänvisningar med ``Any``

````python
from typing import Any

def describe_person(person: Dict[str, Any]) -> str:
    """Returnerar en beskrivning av en person."""
    return f"{person.get('name', 'Okänd')} är {person.get('age', 'okänd ålder')} år gammal."
````

````python
person_info = {"name": "Alice", "age": 30}
description = describe_person(person_info)
print(f"Personbeskrivning: {description}")
````
