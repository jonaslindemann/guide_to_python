### Matematiska algoritmer

```python
import math

# Största gemensamma delaren
a, b = 48, 18
sgd = math.gcd(a, b)
print(f"SGD({a}, {b}) = {sgd}")

# Minsta gemensamma multipeln
mgm = abs(a * b) // sgd
print(f"MGM({a}, {b}) = {mgm}")

# Fakultet
n = 5
fakultet = math.factorial(n)
print(f"{n}! = {fakultet}")

# Kombinatorik (från Python 3.8+)
try:
    # Kombinationer: C(n,k) = n! / (k!(n-k)!) 
    n, k = 10, 3
    kombinationer = math.comb(n, k)
    print(f"C({n},{k}) = {kombinationer}")

    # Permutationer: P(n,k) = n! / (n-k)!
    permutationer = math.perm(n, k)
    print(f"P({n},{k}) = {permutationer}")
except AttributeError:
    print("Kombinatorikfunktioner kräver Python 3.8+")
```

### Statistiska funktioner

```python
import statistics

data = [1.2, 2.3, 3.1, 1.8, 2.9, 3.4, 2.1, 1.9, 2.7, 3.2]

# Centralmått
medelvärde = statistics.mean(data)
median = statistics.median(data)
try:
    typvärde = statistics.mode(data)
except statistics.StatisticsError:
    typvärde = "Inget entydigt typvärde"

print(f"Medelvärde: {medelvärde:.2f}")
print(f"Median: {median:.2f}")
print(f"Typvärde: {typvärde}")

# Spridningsmått
standardavvikelse = statistics.stdev(data)
varians = statistics.variance(data)

print(f"Standardavvikelse: {standardavvikelse:.3f}")
print(f"Varians: {varians:.3f}")

# Kvartiler
q1 = statistics.quantiles(data, n=4)[0]  # Första kvartilen
q3 = statistics.quantiles(data, n=4)[2]  # Tredje kvartilen
print(f"Q1: {q1:.2f}, Q3: {q3:.2f}")
```

### Använda temporära filer

```python
import tempfile
import os

temp_fd, temp_path = tempfile.mkstemp()
print('Temporär fil', temp_path, 'skapad.')
print('Är det en faktisk fil:', os.path.isfile(temp_path))
```

```python
try:
    with open(temp_fd, 'w+t') as temp_file:
        temp_file.write('this is written to the temp file')
        temp_file.seek(0)
        print(temp_file.read())
finally:
    os.remove(temp_path)

print(os.path.isfile(temp_path))
```

```python
with tempfile.TemporaryFile('w+t') as temp_file:
    print('Temporär fil', temp_file.name, 'skapad.')
    print(os.path.isfile(temp_file.name))
    temp_file.write('this is written to the temp file')
    temp_file.seek(0)
    print(temp_file.read())

print(os.path.isfile(temp_file.name))
```

```python
import subprocess, time

p = subprocess.Popen(['tracert', '-h', '2', 'www.google.se'])

while p.poll() is None:
    print('Väntar...')
    time.sleep(1)

if p.returncode == 0:
    print('Processen returnerade 0')
else:
    print('Processen returnerade felkoden = ', p.returncode)
```

### Fånga utdata från ett startat program

```python
import subprocess, time

p = subprocess.Popen(['tracert', '-h', '2', 'www.google.se'], stdout=subprocess.PIPE, stderr=subprocess.PIPE, universal_newlines=True)

while p.poll() is None:
    print('Väntar...')
    time.sleep(1)

stdout, stderr = p.communicate()

if p.returncode == 0:
    print('Processen returnerade 0')
    print('standard output:')
    print(stdout)
    print('standard error:')
    print(stderr)
else:
    print('Processen returnerade felkoden = ', p.returncode)
```

## Dataarkivering och komprimering

### Komprimering av dataströmmar

```python
import pickle, zlib

my_data = {"a number": 42, "a list": list(range(1000)), "a dict": {'a': 1, 'b': 2}}

my_data_dump = pickle.dumps(my_data)
print(len(my_data_dump))
```

```python
my_data_compressed = zlib.compress(my_data_dump)
print(len(my_data_compressed))
```

```python
my_data_uncompressed = zlib.decompress(my_data_compressed)

my_data_copy = pickle.loads(my_data_uncompressed)
print(my_data_copy)
```

### Filöverföring med FTP – ftplib

```python
import ftplib

with ftplib.FTP("ftp.acc.umu.se") as ftp:
    ftp = ftplib.FTP("ftp.acc.umu.se")
    ftp.login()
    print(ftp.getwelcome())
    print(ftp.dir())
    #print(ftp.nlst())
    print(ftp.mlsd())
    print(ftp.pwd())
```

```python
import ftplib
import cdflib


with ftplib.FTP_TLS() as ftp:
    ftp.connect("nssdcftp.gsfc.nasa.gov", 21)
    ftp.login("anonymous", "")
    print(ftp.getwelcome())
    print(ftp.pwd())

    ftp.cwd("/pub/data/image/hk/ads_k0/2005")
    # print(ftp.pwd())

    file_list = ftp.nlst()
    print(len(file_list))

    image_filename = file_list[-1]

    print("Downloading", image_filename)

    with open(image_filename, 'wb') as image_file:
        ftp.retrbinary('RETR '+image_filename, image_file.write)

cdf_file = cdflib.CDF(image_filename)
file_info = cdf_file.cdf_info()
for item in file_info.zVariables:
    print(item)
```
