### Gruppering och aggregering

```python
equipment_ids = ['Pump_001', 'Motor_002', 'Tank_003', 'Compressor_004', 'Valve_005']

large_series_idx = np.random.choice(equipment_ids, size=100)
temperatures = pd.Series(np.random.uniform(70, 90, size=100), index=large_series_idx)
pressures = pd.Series(np.random.uniform(0.5, 10, size=100), index=large_series_idx)
vibrations = pd.Series(np.random.uniform(0.1, 3, size=100), index=large_series_idx)

equipment_df = pd.DataFrame({
    'temperature_c': temperatures,
    'pressure_bar': pressures,
    'vibration_mm_s': vibrations
})

print(equipment_df.head(10))
```

```python
grouped_df = equipment_df.groupby(equipment_df.index).mean()
print(grouped_df)   
```

```python
pressure_max_df = equipment_df.groupby(equipment_df.index)['pressure_bar'].max()
print(pressure_max_df)
```

```python
agg_df = equipment_df.groupby(equipment_df.index).agg(['mean', 'std'])
print(agg_df)
```

### Kombinera, slå ihop och slå samman

```python
equipment_master = pd.DataFrame({
    'equipment_id': ['P001', 'P002', 'T001', 'T002'],
    'equipment_name': ['Main Feed Pump', 'Backup Pump', 'Reactor Tank', 'Storage Tank'],
    'installation_date': pd.to_datetime(['2020-01-15', '2020-02-01', '2019-06-10', '2019-07-20']),
    'manufacturer': ['Grundfos', 'Grundfos', 'Pfaudler', 'Pfaudler']
})

print(equipment_master)
```

## Exempel - Väderdata

```python
import pandas as pd
import numpy as np

from weather_data import *
```

```python
# Ladda ner värden från olika väderstationer

station_ids = [
    'USW00094728_NYC',  # New York City
    'USW00012839_MIA',  # Miami
    'USW00023062_DEN'   # Denver
]

# Ange start- och slutdatum för datan

start_date = '2020-01-01'
end_date = '2023-12-31'

# Ladda ner och rensa väderdata

weather_df = download_weather_data(station_ids, start_date, end_date)
```

```python
print(weather_df.head(10))
```

```python
print(weather_df.dtypes)
```

```python
print(weather_df.shape)
```

### Temperatur per stad och år (Pivot-tabell)

```python
temp_pivot = weather_df.pivot_table(
    values='TAVG', 
    index='YEAR', 
    columns='CITY', 
    aggfunc='mean'
).round(1)

print(temp_pivot)
```

### Gruppera efter månad och stad

```python
monthly_stats_mean = weather_df.groupby(['CITY']).mean(numeric_only=True).round(2)
print(monthly_stats_mean)
print()
```

```python
monthly_stats_aggregate = weather_df.groupby(['CITY', 'MONTH']).agg({
    'TAVG': ['mean', 'std'],
    'PRCP': ['mean', 'sum'],
    'TMAX': 'max',
    'TMIN': 'min',
    'AWND': 'mean'
}).round(2)
```

```python
yearly_stats_aggregate = weather_df.groupby(['CITY', 'YEAR']).agg({
    'TAVG': ['mean', 'std'],
    'PRCP': ['mean', 'sum'],
    'TMAX': 'max',
    'TMIN': 'min',
    'AWND': 'mean'
}).round(2)

print(yearly_stats_aggregate)
```

### Kombinera temperaturdata med väderdata (merge)

```python
climate_zones = pd.DataFrame({
    'CITY': ['NYC', 'MIA', 'DEN'],
    'CLIMATE_ZONE': ['Humid Continental', 'Tropical', 'Semi-Arid'],
    'ELEVATION_FT': [33, 6, 5280],
    'LATITUDE': [40.7, 25.8, 39.7],
    'LONGITUDE': [-74.0, -80.3, -105.0]
})

print("Klimatzoninformation:")

print(climate_zones)
print()
```

```python
weather_enhanced = weather_df.merge(climate_zones, on='CITY', how='left')
print(weather_enhanced)
```

### Slå samman information (concat)

```python
annual_summaries = []
for year in weather_df['YEAR'].unique():

    # Skapa tabell med data för year

    year_data = weather_df[weather_df['YEAR'] == year]
```

```python
    summary = year_data.groupby('CITY').agg({
        'TAVG': 'mean',
        'PRCP': 'sum',
        'TMAX': 'max',
        'TMIN': 'min',
        'AWND': 'mean'
    }).round(2)    
```

```python
    summary['YEAR'] = year
    summary['SEASON_TEMP_RANGE'] = summary['TMAX'] - summary['TMIN']
    annual_summaries.append(summary)
```

```python
combined_annual = pd.concat(annual_summaries, ignore_index=False)
print(combined_annual)    
```
