## Flera diagram i samma figurfönster

### Grundläggande användning av plt.subplots()

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0.0, 2*np.pi, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

# Skapa figur med 2 rader, 1 kolumn
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6))

# Plotta i första subploten
ax1.plot(x, y1, 'r-')
ax1.set_xlabel("x")
ax1.set_ylabel("sin(x)")
ax1.set_title("Sinusfunktion")

# Plotta i andra subploten
ax2.plot(x, y2, 'b-')
ax2.set_xlabel("x")
ax2.set_ylabel("cos(x)")
ax2.set_title("Cosinusfunktion")

# Justera layout för att undvika överlappning
plt.tight_layout()
plt.show()
```

### Flera diagram i rutnät

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)

# Skapa 2x2 rutnät av subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Plotta i varje subplot med indexering
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title('sin(x)')

axes[0, 1].plot(x, np.cos(x))
axes[0, 1].set_title('cos(x)')

axes[1, 0].plot(x, np.sin(2*x))
axes[1, 0].set_title('sin(2x)')

axes[1, 1].plot(x, np.cos(2*x))
axes[1, 1].set_title('cos(2x)')

# Lägg till gemensamma axeletiketter
for ax in axes.flat:
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.grid(True)

plt.tight_layout()
plt.show()
```

### Dela axlar mellan subplots

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Skapa subplots som delar x-axel
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(8, 6))

ax1.plot(x, y1, 'r-')
ax1.set_ylabel('sin(x)')
ax1.grid(True)

ax2.plot(x, y2, 'b-')
ax2.set_xlabel('x')
ax2.set_ylabel('cos(x)')
ax2.grid(True)

plt.tight_layout()
plt.show()
```

### Äldre syntax med plt.subplot()

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0.0, 2*np.pi, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(1)

plt.subplot(211)  # 2 rader, 1 kolumn, subplot 1
plt.xlabel("x")
plt.ylabel("y1")
plt.plot(x, y1, 'r-')

plt.subplot(212)  # 2 rader, 1 kolumn, subplot 2
plt.xlabel("x")
plt.ylabel("y2")
plt.plot(x, y2, 'b-')

plt.tight_layout()
plt.show()
```

## Andra typer av diagram

### Histogram

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

# Exempeldata
mu = 100    # Distributionens medelvärde
sigma = 15  # Standardavvikelse

# Skapa sekvens av 10 000 värden
x = mu + sigma * np.random.randn(10000)

num_bins = 50

plt.hist(x, num_bins)
plt.title('Histogram')
plt.xlabel('x')
plt.ylabel('Frekvens')

plt.show()
```

### Plottning av punkter med värden

```python
import matplotlib.pyplot as plt
import numpy as np

n = 200

x1, y1 = np.random.rand(2, n)
x2, y2 = np.random.rand(2, n)
x3, y3 = np.random.rand(2, n)

plt.scatter(x1, y1, c='red')
plt.scatter(x2, y2, c='green')
plt.scatter(x3, y3, c='blue')

plt.title('Plottning av punkter')
plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

n = 200

x1, y1 = np.random.rand(2, n)
x2, y2 = np.random.rand(2, n)
x3, y3 = np.random.rand(2, n)

s1 = 200.0 * np.random.rand(n)
s2 = 200.0 * np.random.rand(n)
s3 = 200.0 * np.random.rand(n)

plt.scatter(x1, y1, c='red', s=s1, label='Serie 1', alpha=0.3, edgecolors='none')
plt.scatter(x2, y2, c='green', s=s2, label='Serie 2', alpha=0.3, edgecolors='none')
plt.scatter(x3, y3, c='blue', s=s3, label='Serie 3', alpha=0.3, edgecolors='none')

plt.title('Plottning av punkter')
plt.legend()
plt.grid(True)

plt.show()
```

## Rekommenderat arbetssätt med Matplotlib

### När ska man använda pyplot vs objektorienterat gränssnitt?

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)

# Skapa figur och axel-objekt
fig, ax = plt.subplots(figsize=(8, 6))

# Använd axel-objektets metoder
ax.plot(x, np.sin(x), label='sin(x)')
ax.plot(x, np.cos(x), label='cos(x)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Trigonometriska funktioner')
ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```

### Hantera figurfönster korrekt

### Använd beskrivande variabelnamn

```python
import matplotlib.pyplot as plt
import numpy as np

# Dåligt - otydliga namn
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x, y1)
plt.plot(x, y2)

# Bra - beskrivande namn
tid = np.linspace(0, 10, 100)
position = np.sin(tid)
hastighet = np.cos(tid)
plt.plot(tid, position, label='Position')
plt.plot(tid, hastighet, label='Hastighet')
plt.legend()
```

### Använd gridlinjer med måtta

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Standard gridlinjer
ax1.plot(x, y)
ax1.grid(True)
ax1.set_title('Standard grid')

# Anpassade gridlinjer (mer diskreta)
ax2.plot(x, y)
ax2.grid(True, alpha=0.3, linestyle='--', linewidth=0.5)
ax2.set_title('Diskret grid')

plt.tight_layout()
plt.show()
```

### Färgval och tillgänglighet

```python
import matplotlib.pyplot as plt
import numpy as np

# Använd färgscheman som är färgblindvänliga
x = np.linspace(0, 10, 100)

fig, ax = plt.subplots(figsize=(8, 6))

# Färger från en färgblindvänlig palett
ax.plot(x, np.sin(x), color='#0173B2', label='Serie 1', linewidth=2)
ax.plot(x, np.cos(x), color='#DE8F05', label='Serie 2', linewidth=2)
ax.plot(x, np.sin(2*x), color='#029E73', label='Serie 3', linewidth=2)

ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

### Optimera prestanda för stora dataset

```python
import matplotlib.pyplot as plt
import numpy as np

# Stort dataset
x = np.linspace(0, 100, 1000000)
y = np.sin(x)

# Metod 1: Nedsampling av data
downsample = 1000
fig, ax = plt.subplots()
ax.plot(x[::downsample], y[::downsample])
ax.set_title('Nedsamplade data')
plt.show()
plt.close()

# Metod 2: Använd rasterisering för många punkter
fig, ax = plt.subplots()
ax.plot(x, y, rasterized=True)
ax.set_title('Rasteriserad plot')
plt.savefig('large_plot.pdf', dpi=150)
plt.close()
```
