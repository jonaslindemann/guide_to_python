### Egenskaper (properties)

```python
class Point:
    def __init__(self, x=0.0, y=0.0):
        self.__x = x
        self.__y = y

    def set_x(self, x):
        self.__x = x

    def set_y(self, y):
        self.__y = y

    def set(self, x, y):
        self.__x = x
        self.__y = y

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    x = property(get_x, set_x)
    y = property(get_y, set_y)
```

```python
p = Point()

p.x = 42.0
p.y = 84.0
print(p.x)
print(p.y)
```

### Egenskaper med hjälp av dekoratörer (@property)

```python
class Point:
    def __init__(self, x=0.0, y=0.0):
        self.__x = x
        self.__y = y

    @property
    def x(self):
        return self.__x

    @property
    def y(self):
        return self.__y
```

```python
class Point:
    def __init__(self, x=0.0, y=0.0):
        self.__x = x
        self.__y = y

    @property
    def x(self):
        return self.__x

    @x.setter
    def x(self, x):
        self.__x = x

    @property
    def y(self):
        return self.__y

    @y.setter
    def y(self, y):
        self.__y = y
```

```python
class Circle(Point):
    def __init__(self, x=0.0, y=0.0, r=1.0):
        super().__init__(x, y)
        self._r = r
```

```python
class Circle(Point):
    def __init__(self, x=0.0, y=0.0, r=1.0):
        super().__init__(x, y)
        self.__r = r

    def set_r(self, r):
        self.__r = r

    def get_r(self):
        return self.__r

    def copy_from(self, c):
        super().copy_from(c)
        self.r = c.r

    def __str__(self):
        return "Circle("+str(self.x)+", "+str(self.y)+", "+str(self._r)+")"

    r = property(get_r, set_r)
```

```python
p = Point(1.0, 2.0)
c = Circle(2.0, 4.0, 8.0)

c.r = 10.0

p.move(1.0, 1.0)
c.move(1.0, 1.0)

print(p)
print(c)
```

## Basklasser

```python
class Base:
    def __init__(self, name="noname"):
        self._name = name

    def set_name(self, name):
        self._name = name

    def get_name(self):
        return self._name

    def copy_from(self, o):
        self._name = o.name

    def __str__(self):
        return self._name+" - Base() "

    name = property(get_name, set_name)
```

### Metoder för egna datastrukturer

```python
class Container:
    def __init__(self):
        self.__items = []

    def __len__(self):
        print("(__len__) Returnerar längden av containern")
        return len(self.__items)

    def __getitem__(self, index):
        print(f"(__getitem__) Hämtar värdet vid index {index}")
        return self.__items[index]

    def __setitem__(self, index, value):
        print(f"(__setitem__) Sätter värdet vid {index} till {value}")
        self.__items[index] = value

    def __contains__(self, item):
        print(f"(__contains__) Undersöker om {item} finns i containern")
        return item in self.__items

    def append(self, item):
        print(f"Lägger till: {item}")
        self.__items.append(item)

    def __str__(self):
        return "Container with items: " + str(self.__items)
```
