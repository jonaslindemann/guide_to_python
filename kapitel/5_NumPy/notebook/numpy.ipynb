## Matrismultiplikation och linjär algebra

### Grundläggande matrisoperationer

```python
import numpy as np

A = np.random.rand(4, 4)  # Skapa en 4x4 matris med slumpmässiga värden
print("Matris A:")
print(A)

B = np.random.rand(4, 4)  # Skapa en annan 4x4 matris
print("Matris B:")
print(B)

# Matrismultiplikation med @ (rekommenderat)
print("\nMatrismultiplikation A @ A:")
print(A @ B)

# Alternativt med .dot() (äldre syntax)
print("\nSamma resultat med .dot():")
print(A.dot(B))

# Jämför med elementvis multiplikation
print("\nElementvis multiplikation A * A:")
print(A * B)
```

### Grundläggande linjär algebra operationer

```python
A = np.array([[4, 2, 1],
              [2, 5, 3],
              [1, 3, 6]], dtype=float)

print("Matris A:")
print(A)
```

```python
print("\nTransponat A.T:")
print(A.T)
```

```python
det_A = np.linalg.det(A)
print(f"\nDeterminant: {det_A:.3f}")
```

```python
inv_A = np.linalg.inv(A)
print("\nInvers av A:")
print(inv_A)

print("\nA @ inv_A:")
print(A @ inv_A)  # Matrismultiplikation av A och dess invers
```

```python
eigenvals, eigenvecs = np.linalg.eig(A)
print(f"\nEgenvärden: {eigenvals}")
```

```python
for i, vec in enumerate(eigenvecs.T):
    print(f"Egenvektor {i+1}: {vec}")
```

```python
for i, vec in enumerate(eigenvecs.T):
    lambda_i = eigenvals[i]
    print(f"\nKontroll av egenvektor {i+1}:")
    print("A @ v:", A @ vec)
    print("λ * v:", lambda_i * vec)
    print("Skillnad:", A @ vec - lambda_i * vec)
```

### Lösa linjära ekvationssystem

```python
A = np.array([[2, 1, -1],
              [-3, -1, 2],
              [-2, 1, 2]], dtype=float)

b = np.array([8, -11, -3], dtype=float)

print("Koefficientsmatris A:")
print(A)
print(f"\nHögerled b: {b}")
```

```python
x = np.linalg.solve(A, b)
print(f"\nLösning x: {x}")
```

```python
verification = A @ x
print(f"Verifiering A @ x: {verification}")
print(f"Skillnad från b: {np.abs(verification - b)}")
```

## Generering av slumptal

### Skapa en slumptalsgenerator

```python
rng = np.random.default_rng(seed=42)  # Reproducerbara resultat
```

```python
# Uniform fördelning mellan 0 och 10
uniform_data = rng.uniform(0, 10, 100) 
# normalfördelning med medelvärde 100 och standardavvikelse 15
normal_data = rng.normal(100, 15, 100) 
# Exponentialfördelning med lambda=2
# (medelvärde 1/lambda)
# (standardavvikelse 1/lambda)
exponential_data = rng.exponential(2, 100)

print("Uniform fördelning (0-10):")
print(f"Medel: {np.mean(uniform_data):.2f}, Std: {np.std(uniform_data):.2f}")

print("Normal fördelning (μ=100, σ=15):")
print(f"Medel: {np.mean(normal_data):.2f}, Std: {np.std(normal_data):.2f}")

print("Exponentialfördelning (λ=2):")
print(f"Medel: {np.mean(exponential_data):.2f}, Std: {np.std(exponential_data):.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ... slumptalsgenerator och data som tidigare ...

# Plotta histogram för varje fördelning

plt.figure(figsize=(12, 8))
plt.subplot(3, 1, 1)   
plt.hist(uniform_data, bins=30, color='blue', alpha=0.7)
plt.title('Uniform fördelning (0-10)')
plt.subplot(3, 1, 2)
plt.hist(normal_data, bins=30, color='green', alpha=0.7)
plt.title('Normal fördelning (μ=100, σ=15)')
plt.subplot(3, 1, 3)
plt.hist(exponential_data, bins=30, color='red', alpha=0.7)
plt.title('Exponential fördelning (λ=0.5)')
plt.tight_layout()
plt.show()
```

## Praktiska verktyg och rekommendationer

### Prestanda och optimering

```python
# Vektorisering istället för loopar
# Dåligt: Loop i Python
def slow_computation(arr):
    result = np.zeros_like(arr)
    for i in range(len(arr)):
        result[i] = arr[i] ** 2 + 2 * arr[i] + 1
    return result

# Bra: Vektoriserat uttryck
def fast_computation(arr):
    return arr**2 + 2*arr + 1
```

```python
import numpy as np
import timeit

large_array = np.random.random(1000000)

# Mät tid för långsam beräkning
execution_time = timeit.timeit(lambda: slow_computation(large_array), number=10)

print(f"Långsam beräkning tog {execution_time:.4f} sekunder")

# Mät tid för snabb beräkning
execution_time = timeit.timeit(lambda: fast_computation(large_array), number=10)

print(f"Snabb beräkning tog {execution_time:.4f} sekunder")
```
